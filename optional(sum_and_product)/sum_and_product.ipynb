# Optional task: about sum and product

There are two whole numbers:

1 < a, b < 100

One scientist ("Sum") gets provided with sum of numbers,  
another ("Prod") gets provided with product of numbers.

Both scientists know that numbers 1 < a, b < 100.  
Determine the numbers being based on the following dialog:

Prod: I don't know the numbers;  
Sum: I know it;  
Prod: then I know the numbers;  
Sum: then I know the numbers too.

### Function 1

From the problem statement, it is clear that the pair of numbers must produce a non-unique product, meaning that the factorization of the product must not be unique.

```python
def find_pairs_and_products():
    pairs = []
    products = {}
    for a in range(2, 100):
        for b in range(a, 100): 
            pair = (a, b)
            product = a * b
            pairs.append(pair)
            if product not in products:
                products[product] = []
            products[product].append(pair)
    return pairs, products
```

I generate all possible pairs and group them by their products in a dictionary, where the key is the product and the value is a list of pairs. Duplicates are also removed. For example: (3,4) and (4,3) — only one pair is kept.

### Function 2

```python
def find_special_sums(pairs, products):
    sums = {}
    for a, b in pairs:
        total = a + b
        if total not in sums:
            sums[total] = []
        sums[total].append((a, b))
    
    special_sums = []
    for total, pair_list in sums.items():
        all_products = True
        for a, b in pair_list:
            product = a * b
            if len(products[product]) == 1:
                all_products = False
                break
        if all_products:
            special_sums.append(total)
    return special_sums
```

I find sums for which all possible pairs have non-unique products. For each sum, I check all its pairs, and for each pair I compute the product. I verify whether the product is unique using the products dictionary; if a product has multiple pairs, I add it to the special_sums dictionary.


### Function 3

```python
def find_unique_products(products, special_sums):
    unique_products = []
    for product, pair_list in products.items():
        if len(pair_list) > 1:
            pairs_special = 0
            for a, b in pair_list:
                if (a + b) in special_sums:
                    pairs_special += 1
            if pairs_special == 1:
                unique_products.append(product)
    return unique_products
```

I iterate over all products and check whether a product has multiple pairs. Then, for each pair, I check whether its sum belongs to special_sums.


### Function 4

```python
def find_final_pair(pairs, special_sums, unique_products):
    for a, b in pairs:
        total = a + b
        product = a * b
        if total in special_sums and product in unique_products:
            sum_pairs = [(x, y) for x, y in pairs if x + y == total]
            valid_pairs = [(x, y) for x, y in sum_pairs if x * y in unique_products]
            if len(valid_pairs) == 1:
                return (a, b)
    return None
```

I find all pairs (sum_pairs) with the same sum a + b. Among them, I select valid_pairs — pairs whose products are in unique_products. If exactly one pair has a product in unique_products, I return this pair.

### Main part

```python
pairs, products = find_pairs_and_products()
special_sums = find_special_sums(pairs, products)
unique_products = find_unique_products(products, special_sums)
final_pair = find_final_pair(pairs, special_sums, unique_products)

if final_pair:
    x, y = final_pair
    print(f"Found numbers: ({x} і {y})")
else:
    print("I don't know the numbers")
```
