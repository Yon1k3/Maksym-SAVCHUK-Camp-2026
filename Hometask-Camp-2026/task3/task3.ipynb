## Task Description

Four coins (c1, c2, c3, c4) have probabilities of outcome H equal to  
[0.12, 0.27, 0.21, 0.96] respectively. One coin is chosen at random and all
further tests are performed using the same coin.

For the sequence of 9 completed tests:
[H H H T H T H H H],
determine the probability of outcome H in the next flip.

The probability should be recomputed after each test using Bayes’ rule.


## Input data

```python
coins = [0.12, 0.27, 0.21, 0.96]   # P(H) for c1..c4
results = list("HHHTHTHHH")        # 9 flips from the task

# at the start all coins are equally likely
posterior = [0.25, 0.25, 0.25, 0.25]
```

## Bayes update

After each flip result (H or T), we update probabilities of hypotheses
(c1..c4) using Bayes’ rule:

- multiply previous probability by likelihood of the observed result,
- normalize so that probabilities sum to 1.

Then the probability of H in the next flip is computed as:

P(H next) = sum( P(coin_i | data) * P(H | coin_i) )


## Update and forecast functions

```python
def bayes_update(posterior, outcome, coins):
    # update probabilities of coins after one flip result
    new_post = []

    for p_coin, pH in zip(posterior, coins):
        # probability of getting the observed result from this coin
        likelihood = pH if outcome == "H" else (1 - pH)

        # Bayes rule: prior * likelihood
        new_post.append(p_coin * likelihood)

    # normalize so probabilities sum to 1
    s = sum(new_post)
    return [x / s for x in new_post]


def prob_next_H(posterior, coins):
    # compute probability of H in the next flip
    # as a weighted sum over all coins
    return sum(p * pH for p, pH in zip(posterior, coins))
```

## Calculate a list of probabilities after each test

```python
probs = []                     # store probabilities after each test
posterior_now = posterior[:]   # copy initial probabilities

for outcome in results:
    # update coin probabilities using the current flip result
    posterior_now = bayes_update(posterior_now, outcome, coins)

    # compute probability of H in the next flip
    probs.append(prob_next_H(posterior_now, coins))

# round probabilities to two decimal places
probs_2dp = [float(f"{x:.2f}") for x in probs]

print("Probabilities for next flip after each test:")
print(probs_2dp)
```
