#### Energy Engineering and Power Technology

```python
def answer_one():
    energy = pd.read_excel(
        "Energy Indicators.xls",
        header=None,
        skiprows=18,
        skipfooter=38,
        na_values='...'
    )

    # Видаляємо перші два непотрібні стовпці
    energy = energy.iloc[:, 2:6]

    # Перейменовуємо стовпці
    energy.columns = ['Country', 'Energy Supply', 'Energy Supply per Capita', '% Renewable']
    
    # Переводимо Energy Supply у гігаджоулі
    energy['Energy Supply'] = energy['Energy Supply'] * 1000000

    # Прибираємо числа з назв країн
    energy['Country'] = energy['Country'].str.replace(r'\d+', '', regex=True)

    # Прибираємо текст у дужках
    energy['Country'] = energy['Country'].str.replace(r'\s*\(.*\)', '', regex=True)

    # Перейменовуємо країни згідно умови
    energy['Country'] = energy['Country'].replace({
        'Republic of Korea': 'South Korea',
        'United States of America': 'United States',
        'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
        'China, Hong Kong Special Administrative Region': 'Hong Kong'
    })

    # Робимо Country індексом
    energy = energy.set_index('Country')

    # Завантаження та очищення GDP
    GDP = pd.read_csv(
        "API_NY.GDP.MKTP.CD_DS2_en_csv_v2_133326.csv",
        skiprows=4
    )

    # Перейменування країн
    GDP['Country Name'] = GDP['Country Name'].replace({
        "Korea, Rep.": "South Korea",
        "Iran, Islamic Rep.": "Iran",
        "Hong Kong SAR, China": "Hong Kong"
    })

    # Робимо Country Name індексом
    GDP = GDP.set_index('Country Name')

    # Залишаємо тільки роки 2006–2015
    GDP = GDP.loc[:, '2006':'2015']

    # Завантаження ScimEn
    ScimEn = pd.read_excel("scimagojr country rank 1996-2024.xlsx")

    # Беремо тільки топ-15 країн
    ScimEn = ScimEn[ScimEn['Rank'] <= 15]

    # Залишаємо тільки потрібні стовпці
    ScimEn = ScimEn[['Rank', 'Country', 'Documents', 'Citable documents',
                     'Citations', 'Self-citations', 'Citations per document',
                     'H index']]

    # Робимо Country індексом
    ScimEn = ScimEn.set_index('Country')

    # Об’єднання трьох таблиць
    Top15 = ScimEn.join(energy, how='inner').join(GDP, how='inner')

    return Top15
display(answer_one())
```

#### Question 2

What is the average GDP over the last 10 years for each country? (exclude missing values from this calculation.)

This function should return a Series named avgGDP with 15 countries and their average GDP sorted in descending order.

```python
def answer_two():
    Top15 = answer_one()
    
    # Середній GDP за 2006–2015 для кожної країни
    avgGDP = Top15.loc[:, '2006':'2015'].mean(axis=1)
    
    # Сортуємо за спаданням
    avgGDP = avgGDP.sort_values(ascending=False)
    
    return avgGDP
display(answer_two())
```

#### Question 3

By how much had the GDP changed over the 10 year span for the country with the 6th largest average GDP?

This function should return a single number.

```python
def answer_three():
    Top15 = answer_one()
    
    # Обчислюємо середній GDP
    avgGDP = Top15.loc[:, '2006':'2015'].mean(axis=1)
    
    # Сортуємо за спаданням
    avgGDP = avgGDP.sort_values(ascending=False)
    
    # Беремо країну з 6-м найбільшим середнім GDP
    country = avgGDP.index[5]
    
    # Обчислюємо зміну GDP (2015 - 2006)
    result = Top15.loc[country, '2015'] - Top15.loc[country, '2006']
    
    return result
display(answer_three())
```

#### Question 4

Create a new column that is the ratio of Self-Citations to Total Citations. What is the maximum value for this new column, and what country has the highest ratio?

This function should return a tuple with the name of the country and the ratio.

```python
def answer_four():
    Top15 = answer_one()
    
    # Створюємо новий показник (ratio)
    ratio = Top15['Self-citations'] / Top15['Citations']
    
    # Знаходимо країну з максимальним значенням
    country = ratio.idxmax()
    
    # Максимальне значення
    max_value = ratio.max()
    
    # Повертаємо результат як кортеж
    return (country, max_value)
display(answer_four())
```

Create a column that estimates the population using Energy Supply and Energy Supply per capita. What is the third most populous country according to this estimate?

This function should return a single string value.

```python
def answer_five():
    Top15 = answer_one()
    
    # Оцінка населення
    population = Top15['Energy Supply'] / Top15['Energy Supply per Capita']
    
    # Сортуємо за спаданням
    population = population.sort_values(ascending=False)
    
    # Беремо третю країну
    third_country = population.index[2]
    
    return third_country
display(answer_five())
```

#### Question 6

Create a column that estimates the number of citable documents per person. What is the correlation between the number of citable documents per capita and the energy supply per capita? Use the .corr() method, (Pearson's correlation).

This function should return a single number.

```python
def answer_six():
    Top15 = answer_one()
    
    # Оцінка населення
    population = Top15['Energy Supply'] / Top15['Energy Supply per Capita']
    
    # Кількість цитованих документів на душу населення
    citable_docs_per_capita = Top15['Citable documents'] / population
    
    # Кореляція з Energy Supply per Capita
    correlation = citable_docs_per_capita.corr(Top15['Energy Supply per Capita'])
    
    return correlation
display(answer_six())
```

#### Question 7

Use the following dictionary to group the Countries by Continent, then create a dateframe that displays the sample size (the number of countries in each continent bin), and the sum, mean, and std deviation for the estimated population of each country.

```python
def answer_seven():
    Top15 = answer_one().copy()
    
    # Словник континентів
    ContinentDict  = {
        'China':'Asia', 
        'United States':'North America', 
        'Japan':'Asia', 
        'United Kingdom':'Europe', 
        'Russian Federation':'Europe', 
        'Canada':'North America', 
        'Germany':'Europe', 
        'India':'Asia',
        'France':'Europe', 
        'South Korea':'Asia', 
        'Italy':'Europe', 
        'Spain':'Europe', 
        'Iran':'Asia',
        'Australia':'Australia', 
        'Brazil':'South America'
    }
    
    # Оцінка населення
    Top15['Population'] = Top15['Energy Supply'] / Top15['Energy Supply per Capita']
    
    # Додаємо континент
    Top15['Continent'] = Top15.index.map(ContinentDict)
    
    # Групування і обчислення статистики
    result = Top15.groupby('Continent')['Population'].agg(['size', 'sum', 'mean', 'std'])
    
    return result
display(answer_seven())
```
