## Phase 2: Building a Data Lake with S3, Glue and Athena

### 2.1 S3: Data Storage

An S3 bucket named `data-lake-maksym-2026` was created to store the source data for the Data Lake pipeline.

A CSV file named `sales_data.csv` was uploaded to the bucket. 

![Uploaded CSV file in S3 bucket](images/phase2.1.png)

### 2.2 IAM Role: Access for AWS Glue

An IAM role named `glue-s3-role-maksym` was created for AWS Glue.

This role allows AWS Glue to access the uploaded CSV file in the S3 bucket and create metadata in the Glue Data Catalog.  
The following policies were attached to the role:

- `AmazonS3FullAccess`
- `AWSGlueServiceRole`

This role will be used later by the Glue Crawler to scan the S3 data and automatically create a table schema.

### 2.3 AWS Glue Crawler and Data Catalog

An AWS Glue Crawler named `sales-data-crawler` was created to scan the CSV file stored in the S3 bucket.

The crawler used the IAM role `glue-s3-role-maksym` and the target database `sales_data_db`. After running the crawler, AWS Glue automatically detected the schema of the uploaded file and created a table in the Glue Data Catalog.

The resulting table was named `data_lake_maksym_2026`. The schema includes the following columns:
- `order_id` — `bigint`
- `customer_name` — `string`
- `product` — `string`
- `quantity` — `bigint`
- `price` — `bigint`
- `order_date` — `string`


![Glue Data Catalog table schema](images/phase2.3.png)

### 2.4 Amazon Athena: Querying Data

Amazon Athena was used to query the table created by AWS Glue Crawler.

The database `sales_data_db` and the table `data_lake_maksym_2026` were selected in the Athena Query Editor.  
The following SQL query was executed:

```sql
SELECT *
FROM sales_data_db.data_lake_maksym_2026
LIMIT 10;

![Athena query results](images/phase2.4.png)

## Phase 3: LocalStack and AWS CLI (Optional)

As an optional part of the task, LocalStack was used to simulate AWS services locally through Docker and AWS CLI.

LocalStack was started in a Docker container using the image `localstack/localstack:4.4.0`.  
After the service became ready, AWS CLI commands were executed with the LocalStack endpoint:

```bash
aws --endpoint-url=http://localhost:4566 s3 mb s3://test-bucket
aws --endpoint-url=http://localhost:4566 s3 cp data.csv s3://test-bucket
aws --endpoint-url=http://localhost:4566 s3 ls s3://test-bucket

![LocalStack S3 commands](images/phase3.1.png)

![LocalStack S3 commands](images/phase3.2.png)
